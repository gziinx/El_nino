```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python

sns.set_theme(style="whitegrid", context="talk")

# ============================================================
# ANOS
# ============================================================

anos = [1982, 1997, 2015, 2023]

# ============================================================
# CORES
# ============================================================

CORES = {
    1982: "#672F8A",
    1997: "#E67E22",
    2015: "#168AAD",
    2023: "#D62828"
}

# ============================================================
# PERÍODO — ALTERE AQUI
# ============================================================

dia_inicio = 1
mes_inicio = 1

dia_fim = 31
mes_fim = 12

# ============================================================
# CALCULA A TEMPERATURA MÍNIMA MÉDIA DO PERÍODO
# ============================================================

medias = []

for ano in anos:

    caminho = f"arquivos/gold/sp_{ano}_0101_1231_gold.csv"

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # ========================================================
    # FILTRO DE DATA
    # ========================================================

    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    periodo = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # ========================================================
    # MÉDIA DA TEMPERATURA MÍNIMA
    # ========================================================

    media = periodo["Temperatura_minima"].mean()

    medias.append({
        "Ano": ano,
        "Temperatura_minima": media
    })

df_medias = pd.DataFrame(medias)

# ============================================================
# GRÁFICO
# ============================================================

fig, ax = plt.subplots(figsize=(12, 7))

barras = ax.bar(
    df_medias["Ano"].astype(str),
    df_medias["Temperatura_minima"],
    color=[CORES[ano] for ano in anos],
    width=0.65
)

# ============================================================
# VALOR EM CIMA DE CADA BARRA
# ============================================================

for barra, valor in zip(
    barras,
    df_medias["Temperatura_minima"]
):

    ax.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + 0.05,
        f"{valor:.1f} °C",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold"
    )

# ============================================================
# TÍTULO E EIXOS
# ============================================================

ax.set_title(
    f"Temperatura mínima média — "
    f"{dia_inicio:02d}/{mes_inicio:02d} a "
    f"{dia_fim:02d}/{mes_fim:02d}",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Ano")

ax.set_ylabel(
    "Temperatura mínima média (°C)"
)

# ============================================================
# AJUSTE DO EIXO Y
# ============================================================

y_min = df_medias["Temperatura_minima"].min()

y_max = df_medias["Temperatura_minima"].max()

ax.set_ylim(
    max(0, y_min - 2),
    y_max + 2
)

# ============================================================
# GRADE
# ============================================================

ax.grid(
    axis="y",
    alpha=0.3
)

ax.grid(
    axis="x",
    visible=False
)

plt.tight_layout()

# ============================================================
# SALVAR — NOME AUTOMÁTICO
# ============================================================

nome_arquivo = (
    f"p2_barras_temperatura_minima_"
    f"{dia_inicio:02d}_{mes_inicio:02d}_"
    f"{dia_fim:02d}_{mes_fim:02d}.png"
)

plt.savefig(
    nome_arquivo,
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
sns.set_theme(style="whitegrid", context="talk")

# ============================================================
# CORES
# ============================================================

COR_2023 = "#672F8A"
COR_2026 = "#ff00be"

# ============================================================
# ARQUIVOS
# ============================================================

a = pd.read_csv(
    "arquivos/gold/sp_2023_0101_1231_gold.csv",
    parse_dates=["Data"]
)

c = pd.read_csv(
    "arquivos/gold/sp_2026_0601_0920_gold.csv",
    parse_dates=["Data"]
)

# ============================================================
# FILTRO — 2023
# 01/06/2023 ATÉ 31/12/2023
# ============================================================

a = a[
    (a["Data"] >= "2023-06-01") &
    (a["Data"] <= "2023-09-20")
].copy()

# ============================================================
# FILTRO — 2026
# 01/06/2026 ATÉ 20/09/2026
# ============================================================

c = c[
    (c["Data"] >= "2026-06-01") &
    (c["Data"] <= "2026-09-20")
].copy()

# ============================================================
# MESES
# ============================================================

meses_nome = {
    6: "Jun",
    7: "Jul",
    8: "Ago",
    9: "Set"
}

mes_ordem = [
    "Jun",
    "Jul",
    "Ago",
    "Set"
]

a["mes_nome"] = a["Mes"].map(meses_nome)
c["mes_nome"] = c["Mes"].map(meses_nome)

# ============================================================
# RESUMO MENSAL — 2023
# ============================================================

resumo_a = (
    a.groupby("mes_nome")["Temperatura_minima"]
    .agg(["mean", "min", "max"])
    .reindex(mes_ordem)
)

# ============================================================
# RESUMO MENSAL — 2026
# ============================================================

resumo_c = (
    c.groupby("mes_nome")["Temperatura_minima"]
    .agg(["mean", "min", "max"])
    .reindex(mes_ordem)
)

# ============================================================
# GRÁFICO
# ============================================================

fig, ax = plt.subplots(
    figsize=(12, 6.5)
)

x = range(len(mes_ordem))

# ============================================================
# 2023
# ============================================================

ax.plot(
    x,
    resumo_a["mean"],
    color=COR_2023,
    marker="o",
    markersize=9,
    linewidth=2.5,
    label="2023 (média)"
)

ax.fill_between(
    x,
    resumo_a["min"],
    resumo_a["max"],
    color=COR_2023,
    alpha=0.15
)

# ============================================================
# 2026
# ============================================================

ax.plot(
    x,
    resumo_c["mean"],
    color=COR_2026,
    marker="o",
    markersize=9,
    linewidth=2.5,
    label="2026 (média)"
)

ax.fill_between(
    x,
    resumo_c["min"],
    resumo_c["max"],
    color=COR_2026,
    alpha=0.15
)

# ============================================================
# EIXOS
# ============================================================

ax.set_xticks(x)

ax.set_xticklabels(
    mes_ordem
)

ax.set_title(
    "Temperatura mínima — Média mensal com faixa (min-máx)",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel("Mês")

ax.set_ylabel(
    "Temperatura mínima (°C)"
)

# ============================================================
# LEGENDA
# ============================================================

ax.legend(
    title="Ano"
)

# ============================================================
# GRADE
# ============================================================

ax.grid(
    axis="y",
    alpha=0.3
)

ax.grid(
    axis="x",
    visible=False
)

# ============================================================
# SALVAR
# ============================================================

plt.tight_layout()

plt.savefig(
    "p9_b_linha_faixa_temp_minima_2023_2026.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
sns.set_theme(style="white", context="talk")

# ============================================================
# CORES
# ============================================================

CORES = {
    2015: "#168AAD",
    2023: "#D62828",
    2026: "#D95F02"
}

# ============================================================
# ANOS
# ============================================================

anos = [2015, 2023, 2026]

# ============================================================
# PERÍODO
# ============================================================

dia_inicio = 1
mes_inicio = 6

dia_fim = 20
mes_fim = 9

# ============================================================
# VARIÁVEIS
# ============================================================

variaveis = [
    "Temperatura_maxima",
    "Temperatura_minima",
    "Velocidade_vento",
    "Soma_chuva"
]

nomes_curtos = [
    "T.Máxima",
    "T.Mínima",
    "Vel. Vento",
    "Chuva"
]

# ============================================================
# PREPARAR OS DADOS
# ============================================================

correlacoes = {}

for ano in anos:

    # ========================================================
    # CAMINHO DOS ARQUIVOS
    # ========================================================

    if ano == 2026:

        caminho = (
            "arquivos/gold/"
            "sp_2026_0601_0920_gold.csv"
        )

    else:

        caminho = (
            "arquivos/gold/"
            f"sp_{ano}_0101_1231_gold.csv"
        )

    # ========================================================
    # LEITURA
    # ========================================================

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # ========================================================
    # FILTRO DE DATA
    # ========================================================

    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    df = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # ========================================================
    # MATRIZ DE CORRELAÇÃO
    # ========================================================

    corr = df[variaveis].corr()

    corr.columns = nomes_curtos
    corr.index = nomes_curtos

    correlacoes[ano] = corr

# ============================================================
# GRÁFICO
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 6)
)

# ============================================================
# HEATMAPS
# ============================================================

for ax, ano in zip(axes, anos):

    sns.heatmap(
        correlacoes[ano],
        annot=True,
        fmt=".2f",
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        ax=ax,
        square=True,
        linewidths=0.5,
        cbar_kws={
            "shrink": 0.8
        }
    )

    ax.set_title(
        str(ano),
        fontsize=18,
        fontweight="bold"
    )

    ax.set_xlabel("")
    ax.set_ylabel("")

# ============================================================
# TÍTULO
# ============================================================

fig.suptitle(
    "Correlação entre variáveis meteorológicas\n"
    "2015 × 2023 × 2026 — Período: 01/06 a 20/09",
    fontsize=17,
    fontweight="bold"
)

# ============================================================
# AJUSTES
# ============================================================

plt.tight_layout(
    rect=[0, 0, 1, 0.92]
)

# ============================================================
# SALVAR
# ============================================================

plt.savefig(
    "p8_c_matriz_correlacao_2015_2023_2026.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```
