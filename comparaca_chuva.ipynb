```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid", context="talk")

# ============================================================
# ANOS
# ============================================================

anos = [1982, 1997, 2015, 2023, 2026]

# ============================================================
# PERÍODO — ALTERE AQUI
# ============================================================

dia_inicio = 1
mes_inicio = 6

dia_fim = 20
mes_fim = 9

# ============================================================
# CORES
# ============================================================

CORES = {
    1982: "#672F8A",
    1997: "#E67E22",
    2015: "#168AAD",
    2023: "#D62828",
    2026: "#D62828"
}

# ============================================================
# CALCULA A CHUVA TOTAL DO PERÍODO
# ============================================================

chuvas = []

for ano in anos:

    # ========================================================
    # CAMINHO DOS ARQUIVOS
    # ========================================================

    if ano == 2026:
        caminho = "arquivos/gold/sp_2026_0601_0920_gold.csv"
    else:
        caminho = f"arquivos/gold/sp_{ano}_0101_1231_gold.csv"

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # ========================================================
    # FILTRO DE DATA
    # ========================================================

    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    df = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # ========================================================
    # CHUVA TOTAL
    # ========================================================

    chuva_total = df["Soma_chuva"].sum()

    chuvas.append({
        "Ano": ano,
        "Chuva_total": chuva_total
    })

df_chuvas = pd.DataFrame(chuvas)

# ============================================================
# GRÁFICO
# ============================================================

fig, ax = plt.subplots(figsize=(12, 7))

barras = ax.bar(
    df_chuvas["Ano"].astype(str),
    df_chuvas["Chuva_total"],
    color=[CORES[ano] for ano in anos],
    width=0.65
)

# ============================================================
# VALOR EM CIMA DE CADA BARRA
# ============================================================

for barra, valor in zip(
    barras,
    df_chuvas["Chuva_total"]
):

    ax.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + 0.05,
        f"{valor:.1f}",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold"
    )

# ============================================================
# TÍTULO E EIXOS
# ============================================================

ax.set_title(
    f"Chuva total — "
    f"{dia_inicio:02d}/{mes_inicio:02d} a "
    f"{dia_fim:02d}/{mes_fim:02d}",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Ano")
ax.set_ylabel("Chuva total")

# ============================================================
# AJUSTE DO EIXO Y
# ============================================================

y_max = df_chuvas["Chuva_total"].max()

ax.set_ylim(
    0,
    y_max * 1.15
)

# ============================================================
# GRADE
# ============================================================

ax.grid(
    axis="y",
    alpha=0.3
)

ax.grid(
    axis="x",
    visible=False
)

plt.tight_layout()

# ============================================================
# SALVAR
# ============================================================

plt.savefig(
    "p3_barras_chuva_total_5_anos.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
sns.set_theme(style="white", context="talk")

# ============================================================
# CORES
# ============================================================

COR_2015 = "#672F8A"
COR_2026 = "#ff00be"

# ============================================================
# ARQUIVOS
# ============================================================

a = pd.read_csv(
    "arquivos/gold/sp_2015_0101_1231_gold.csv",
    parse_dates=["Data"]
)

b = pd.read_csv(
    "arquivos/gold/sp_2026_0601_0920_gold.csv",
    parse_dates=["Data"]
)

# ============================================================
# FILTRO DE DATA
# ============================================================

a = a[
    (a["Data"] >= "2015-06-01") &
    (a["Data"] <= "2015-09-20")
].copy()

b = b[
    (b["Data"] >= "2026-06-01") &
    (b["Data"] <= "2026-09-20")
].copy()

# ============================================================
# VARIÁVEIS
# ============================================================

variaveis = [
    "Temperatura_media",
    "Temperatura_maxima",
    "Temperatura_minima",
    "Velocidade_vento",
    "Soma_chuva"
]

labels_radar = [
    "Temp.\nMédia",
    "Temp.\nMáxima",
    "Temp.\nMínima",
    "Vel.\nVento",
    "Chuva\nTotal"
]

# ============================================================
# MÉDIAS
# ============================================================

medias_a = a[variaveis].mean()
medias_b = b[variaveis].mean()

# ============================================================
# NORMALIZAÇÃO
# ============================================================

norm_min = pd.concat([
    a[variaveis],
    b[variaveis]
]).min()

norm_max = pd.concat([
    a[variaveis],
    b[variaveis]
]).max()

norm_a = (
    (medias_a - norm_min) /
    (norm_max - norm_min)
).values

norm_b = (
    (medias_b - norm_min) /
    (norm_max - norm_min)
).values

# ============================================================
# ÂNGULOS
# ============================================================

angles = np.linspace(
    0,
    2 * np.pi,
    len(variaveis),
    endpoint=False
).tolist()

norm_a = np.concatenate(
    (norm_a, [norm_a[0]])
)

norm_b = np.concatenate(
    (norm_b, [norm_b[0]])
)

angles += angles[:1]

# ============================================================
# GRÁFICO RADAR
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 9),
    subplot_kw={"projection": "polar"}
)

# ============================================================
# 2015
# ============================================================

ax.plot(
    angles,
    norm_a,
    color=COR_2015,
    linewidth=2,
    label="2015"
)

ax.fill(
    angles,
    norm_a,
    color=COR_2015,
    alpha=0.25
)

# ============================================================
# 2026
# ============================================================

ax.plot(
    angles,
    norm_b,
    color=COR_2026,
    linewidth=2,
    label="2026"
)

ax.fill(
    angles,
    norm_b,
    color=COR_2026,
    alpha=0.25
)

# ============================================================
# EIXOS
# ============================================================

ax.set_xticks(angles[:-1])

ax.set_xticklabels(labels_radar)

ax.set_yticklabels([])

# ============================================================
# TÍTULO
# ============================================================

ax.set_title(
    "Raio-X comparativo — 2015 × 2026",
    fontsize=14,
    fontweight="bold",
    pad=30
)

# ============================================================
# LEGENDA
# ============================================================

ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.3, 1.1),
    title="Ano"
)

# ============================================================
# SALVAR
# ============================================================

plt.tight_layout()

plt.savefig(
    "p10_a_radar_2015_2026.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```
