```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid", context="talk")

# ============================================================
# ANOS
# ============================================================
anos = [1982, 1997, 2015, 2023]

# ============================================================
# PERÍODO — ALTERE AQUI
# ============================================================
dia_inicio = 1
mes_inicio = 1

dia_fim = 31
mes_fim = 12

# ============================================================
# CORES
# ============================================================
CORES = {
    1982: "#672F8A",
    1997: "#E67E22",
    2015: "#168AAD",
    2023: "#D62828"
}

# ============================================================
# CALCULA A VELOCIDADE MÉDIA DO PERÍODO
# ============================================================
medias = []

for ano in anos:

    caminho = f"arquivos/gold/sp_{ano}_0101_1231_gold.csv"

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # ========================================================
    # FILTRO DE DATA
    # ========================================================
    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    df = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # ========================================================
    # MÉDIA DA VELOCIDADE DO VENTO
    # ========================================================
    media = df["Velocidade_vento"].mean()

    medias.append({
        "Ano": ano,
        "Velocidade_vento": media
    })

df_medias = pd.DataFrame(medias)

# ============================================================
# GRÁFICO
# ============================================================
fig, ax = plt.subplots(figsize=(12, 7))

barras = ax.bar(
    df_medias["Ano"].astype(str),
    df_medias["Velocidade_vento"],
    color=[CORES[ano] for ano in anos],
    width=0.65
)

# ============================================================
# VALOR EM CIMA DE CADA BARRA
# ============================================================
for barra, valor in zip(
    barras,
    df_medias["Velocidade_vento"]
):

    ax.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + 0.05,
        f"{valor:.1f}",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold"
    )

# ============================================================
# TÍTULO E EIXOS
# ============================================================
ax.set_title(
    f"Velocidade média do vento — {dia_inicio:02d}/{mes_inicio:02d} a {dia_fim:02d}/{mes_fim:02d}",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel("Ano")
ax.set_ylabel("Velocidade média do vento")

# ============================================================
# AJUSTE DO EIXO Y
# ============================================================
y_min = df_medias["Velocidade_vento"].min()

ax.set_ylim(
    max(0, y_min - 1),
    df_medias["Velocidade_vento"].max() + 1
)

# ============================================================
# GRADE
# ============================================================
ax.grid(
    axis="y",
    alpha=0.3
)

ax.grid(
    axis="x",
    visible=False
)

plt.tight_layout()

# ============================================================
# SALVAR
# ============================================================
plt.savefig(
    "p2_barras_velocidade_media_vento.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
CORES = {
    1982: "#672F8A",
    1997: "#E67E22",
    2015: "#168AAD",
    2023: "#D62828",
    2026: "#ff00be"
}

# ============================================================
# ANOS
# ============================================================

anos = [1982, 1997, 2015, 2023, 2026]

# ============================================================
# PERÍODO
# ============================================================

dia_inicio = 1
mes_inicio = 6

dia_fim = 20
mes_fim = 9

# ============================================================
# DIREÇÕES DO VENTO
# ============================================================

direcoes = [
    "Norte",
    "Nordeste",
    "Leste",
    "Sudeste",
    "Sul",
    "Sudoeste",
    "Oeste",
    "Noroeste"
]

angulos = {
    d: i * (360 / 8)
    for i, d in enumerate(direcoes)
}

theta = np.deg2rad(
    [angulos[d] for d in direcoes]
)

theta = np.append(theta, theta[0])

# ============================================================
# GRÁFICO
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 9),
    subplot_kw={"projection": "polar"}
)

ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)

# ============================================================
# PROCESSAMENTO E PLOTAGEM
# ============================================================

for ano in anos:

    # --------------------------------------------------------
    # CAMINHO DOS ARQUIVOS
    # --------------------------------------------------------

    if ano == 2026:
        caminho = "arquivos/gold/sp_2026_0601_0920_gold.csv"
    else:
        caminho = f"arquivos/gold/sp_{ano}_0101_1231_gold.csv"

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # --------------------------------------------------------
    # FILTRO DE DATA
    # --------------------------------------------------------

    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    df = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # --------------------------------------------------------
    # CONTAGEM DAS DIREÇÕES
    # --------------------------------------------------------

    cont = (
        df["Direcao_cardinal"]
        .value_counts()
        .reindex(direcoes, fill_value=0)
    )

    valores = np.append(
        cont.values,
        cont.values[0]
    )

    # --------------------------------------------------------
    # ROSA DOS VENTOS
    # --------------------------------------------------------

    ax.plot(
        theta,
        valores,
        color=CORES[ano],
        linewidth=2,
        label=str(ano)
    )

    ax.fill(
        theta,
        valores,
        color=CORES[ano],
        alpha=0.15
    )

# ============================================================
# EIXOS
# ============================================================

ax.set_xticks(
    np.deg2rad(
        list(angulos.values())
    )
)

ax.set_xticklabels(direcoes)

# ============================================================
# TÍTULO
# ============================================================

ax.set_title(
    "Direção do vento — 01/06 a 20/09",
    fontsize=14,
    fontweight="bold",
    pad=30
)

# ============================================================
# LEGENDA
# ============================================================

ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.25, 1.1),
    title="Ano"
)

# ============================================================
# SALVAR
# ============================================================

plt.tight_layout()

plt.savefig(
    "p6_a_rosa_ventos_5_anos.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
sns.set_theme(style="white", context="talk")

# ============================================================
# CORES
# ============================================================

CORES = {
    1982: "#672F8A",
    1997: "#E67E22",
    2015: "#168AAD",
    2023: "#D62828",
    2026: "#D95F02"
}

# ============================================================
# ANOS
# ============================================================

anos = [1982, 1997, 2015, 2023, 2026]

# ============================================================
# PERÍODO — ALTERE AQUI
# ============================================================

dia_inicio = 1
mes_inicio = 6

dia_fim = 20
mes_fim = 9

# ============================================================
# VARIÁVEIS
# ============================================================

variaveis = [
    "Temperatura_maxima",
    "Temperatura_minima",
    "Velocidade_vento",
    "Soma_chuva",
    "Direcao_X",
    "Direcao_Y"
]

nomes_curtos = [
    "T.Máxima",
    "T.Mínima",
    "Vel. Vento",
    "Chuva",
    "Direção X",
    "Direção Y"
]

# ============================================================
# PREPARAR OS DADOS
# ============================================================

correlacoes = {}

for ano in anos:

    # ========================================================
    # CAMINHO DOS ARQUIVOS
    # ========================================================

    if ano == 2026:
        caminho = "arquivos/gold/sp_2026_0601_0920_gold.csv"
    else:
        caminho = f"arquivos/gold/sp_{ano}_0101_1231_gold.csv"

    df = pd.read_csv(
        caminho,
        parse_dates=["Data"]
    )

    # --------------------------------------------------------
    # FILTRO DE DATA
    # --------------------------------------------------------

    inicio = pd.Timestamp(
        year=ano,
        month=mes_inicio,
        day=dia_inicio
    )

    fim = pd.Timestamp(
        year=ano,
        month=mes_fim,
        day=dia_fim
    )

    df = df[
        (df["Data"] >= inicio) &
        (df["Data"] <= fim)
    ].copy()

    # --------------------------------------------------------
    # DIREÇÃO DO VENTO
    # --------------------------------------------------------

    # Direcao_vento = ângulo em graus
    angulo_rad = np.deg2rad(
        df["Direcao_vento"]
    )

    # Componentes circulares da direção
    df["Direcao_X"] = np.cos(angulo_rad)
    df["Direcao_Y"] = np.sin(angulo_rad)

    # --------------------------------------------------------
    # MATRIZ DE CORRELAÇÃO
    # --------------------------------------------------------

    corr = df[variaveis].corr()

    corr.columns = nomes_curtos
    corr.index = nomes_curtos

    correlacoes[ano] = corr

# ============================================================
# GRÁFICO
# ============================================================

fig, axes = plt.subplots(
    2,
    3,
    figsize=(19, 12)
)

axes = axes.flatten()

for ax, ano in zip(axes, anos):

    sns.heatmap(
        correlacoes[ano],
        annot=True,
        fmt=".2f",
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        ax=ax,
        square=True,
        cbar_kws={"shrink": 0.8}
    )

    ax.set_title(
        str(ano),
        fontweight="bold"
    )

# ============================================================
# REMOVE O ESPAÇO DO SEXTO GRÁFICO
# ============================================================

axes[5].axis("off")

# ============================================================
# TÍTULO
# ============================================================

fig.suptitle(
    "Correlação entre variáveis climáticas e direção do vento\n"
    "Período: 01/06 a 20/09",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

# ============================================================
# SALVAR
# ============================================================

plt.savefig(
    "p8_c_matriz_correlacao_5_anos.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```
